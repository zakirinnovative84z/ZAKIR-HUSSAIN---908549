```python
#Q:2

numbers = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]

first_five = numbers[:5]
last_four = numbers[-4:]
every_second_reverse = numbers[::-2]

print(first_five)
print(last_four)
print(every_second_reverse)
```

```python
#Q:6

numbers = [10, 20, 30, 40, 50, 60]

k = 2

rotated = numbers[-k:] + numbers[:-k]

print(rotated)
```

```python
#Q:7

numbers = [10, 20, 10, 30, 20, 40, 30, 50]

unique_numbers = []

for num in numbers:
    if num not in unique_numbers:
        unique_numbers.append(num)

print(unique_numbers)
```

```python
#Q:8

data = [[10, 20], [30, 40], [50, 60]]

flattened = []

for sublist in data:
    flattened.extend(sublist)

print(flattened)
```

```python
#Q:9

matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

main_diagonal = [matrix[0][0], matrix[1][1], matrix[2][2]]
secondary_diagonal = [matrix[0][2], matrix[1][1], matrix[2][0]]
middle_column = [matrix[0][1], matrix[1][1], matrix[2][1]]

print("Main Diagonal:", main_diagonal)
print("Main Diagonal Sum:", sum(main_diagonal))

print("Secondary Diagonal:", secondary_diagonal)
print("Secondary Diagonal Sum:", sum(secondary_diagonal))

print("Middle Column:", middle_column)
print("Middle Column Sum:", sum(middle_column))
```

```python
#Q:10

data = [15, 5, 25, 10, 35, 20, 30]

unique_data = sorted(set(data))

largest_three = unique_data[-3:]
smallest_two = unique_data[:2][::-1]

result = largest_three + smallest_two

print(result)
```
